Haz la limpieza de datos nulos o mal codificados, la transformación de variables categóricas, normaliza las variables numéricas.

Los datos nulos están enmascarados como "?" . Al ser variables categóricas se imputarán los nulos con la moda.

```python
for col in ["workclass", "occupation", "native.country"]:
    df[col] = df[col].fillna(df[col].mode()[0])

df.isnull().sum()
```

Transformamos las variables a numéricas

```python
df["income"] = df["income"].apply(lambda x: 1 if x == ">50K" else 0)
df["income"].value_counts()
```

```python
cat_cols = ["workclass", "education", "marital.status", "occupation", 
            "relationship", "race", "sex", "native.country"]

df = pd.get_dummies(df, columns=cat_cols)
df.shape
```

```python
num_cols = ["age", "fnlwgt", "education.num", "capital.gain", "capital.loss", "hours.per.week"]
scaler = MinMaxScaler()
df[num_cols] = scaler.fit_transform(df[num_cols])
df.head()
```

```python
df.drop(columns=["workclass_?"], inplace=True)
```

```python
bool_cols = df.select_dtypes(include="bool").columns
df[bool_cols] = df[bool_cols].astype(int)
df.head()
```

## Paso3. Define el problema de recomendación. 

Plantea cómo vas a estructurar tu sistema de recomendación:

**¿Qué se quiere recomendar?**

Trayectorias socioeconómicas combinando educación, ocupación y tipos de empleo asociados a ingresos mayores a 50K

**¿Cuál será el "usuario" en este caso?**

La persona adulta descrita por las variables demográficas y socioeconómicas

**¿Qué variables definen el perfil de un usuario?**

Edad, educación, ocupación, estado civil, horas trabajadas, sexo, país de origen

## Paso4. Construye el sistema de recomendación. 

Usa uno de los siguientes enfoques:

**Filtrado basado en contenido.** Representa a cada usuario como un vector y calcula similitudes entre usuarios y recomendaciones.

**Filtrado colaborativo.** Simula una matriz de usuarios vs. trayectorias. Aplica k-NN, correlación de Pearson o matrix factorization.

**Sistema híbrido.** Combina ambos enfoques.

```python
X = df.drop(columns=["income"])
y = df["income"]

X_high = X[y == 1].reset_index(drop=True)

def recomendar(perfil_usuario, X_high, top_k=5):
    similitudes = cosine_similarity([perfil_usuario], X_high)[0]
    indices_top = np.argsort(similitudes)[::-1][:top_k]
    perfiles_similares = X_high.iloc[indices_top]
    return perfiles_similares, similitudes[indices_top]
```

## Paso5. Pruebas con casos simulados. 

Construye perfiles simulados de usuarios hipotéticos y fijate qué trayectorias (educación, ocupación, etc.) les recomendaría el sistema para mejorar su ingreso estimado.

```python
perfil_usuario = np.zeros(X.shape[1])

perfil_usuario[X.columns.get_loc("age")] = 25 / 90
perfil_usuario[X.columns.get_loc("education.num")] = 9 / 15
perfil_usuario[X.columns.get_loc("hours.per.week")] = 20 / 99
perfil_usuario[X.columns.get_loc("capital.gain")] = 0
perfil_usuario[X.columns.get_loc("capital.loss")] = 0
perfil_usuario[X.columns.get_loc("fnlwgt")] = 0.1
perfil_usuario[X.columns.get_loc("sex_Male")] = 1
perfil_usuario[X.columns.get_loc("workclass_Private")] = 1
perfil_usuario[X.columns.get_loc("native.country_United-States")] = 1

perfiles_similares, similitudes = recomendar(perfil_usuario, X_high, top_k=5)

print("Similitudes:", similitudes)
perfiles_similares
```
